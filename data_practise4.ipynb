```python
# Introduction to Tuples
```

```python
# Create a Tuple
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple)
```

```python
# Tuple Length
print(len(thistuple))
```

```python
# Access Tuple Items
print(thistuple[1])
```

```python
# Range of Indexes
print(thistuple[2:5])
```

```python
# Change Tuple Values
tuple = list(thistuple)
tuple[1] = "strawberry"
print(tuple)
```

```python
# Loop Through a Tuple
for x in thistuple:
    print(x)
```

```python
# Introduction to Dictionary
```

```python
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}
print(thisdict)
```

```python
# Dictionary Items
print(thisdict["brand"])
```

```python
# Dictionary Length
print(len(thisdict))
```

```python
# Data Types
print(type(thisdict))
```

```python
# Accessing Items
x = thisdict["model"]
print(x)
```

```python
# Get Keys
x = thisdict.keys()
print(x)
```

```python
# Get Values
x = thisdict.values()
print(x)
```

```python
# Get Items
x = thisdict.items()
print(x)
```

```python
# Change Values
thisdict["year"] = 2018
print(thisdict)
```

```python
# Update Dictionary
thisdict.update({"year": 2020})
print(thisdict)
```

```python
# Adding Items
thisdict["color"] = "red"
print(thisdict)
```

```python
# Loop Through a Dictionary
for x in thisdict:
    print(x)
```
